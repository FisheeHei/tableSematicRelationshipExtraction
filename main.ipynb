```python
# 独立清理单元格：只删除 external-libraries 和 data。
# 需要重新切换环境或释放空间时，可以单独运行本单元格。
import os
import shutil
import stat

ROOT = "/home/aistudio"
TARGETS = ["external-libraries", "data"]

def handle_remove_readonly(func, path, exc_info):
    try:
        os.chmod(path, stat.S_IWRITE)
        func(path)
    except Exception:
        pass

for name in TARGETS:
    path = os.path.join(ROOT, name)
    if not os.path.exists(path):
        continue

    try:
        if os.path.isdir(path):
            shutil.rmtree(path, onerror=handle_remove_readonly)
        else:
            os.remove(path)
        print(f"已删除: {path}")
    except Exception as e:
        print(f"删除失败: {path} -> {e}")

print("清理完成")
```
